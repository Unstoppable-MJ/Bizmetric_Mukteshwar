```python
price_list = {
    'book': 50,
    'pen': 10,
    'pencil': 5
}

order = {}

print("Welcome to Stationery Shop")
print("Items available: Book, Pen, Pencil")

while True:
    item = input("Enter item name (book/pen/pencil): ").lower()

    if item not in price_list:
        print("Item not available")
        continue

    qty = int(input("Enter quantity: "))

    if item in order:
        order[item] += qty
    else:
        order[item] = qty

    more = input("Do you want to add more items? (yes/no): ").lower()
    if more != 'yes':
        break

print("\nOrder taken successfully")
print("Delivering your order")

total_amount = 0
bill = {}

for item, qty in order.items():
    amount = price_list[item] * qty
    bill[item] = {
        'quantity': qty,
        'price': price_list[item],
        'amount': amount
    }
    total_amount += amount

print("\nBill generated successfully")

choice = input("Do you want to print e-bill? (yes/no): ").lower()

if choice == 'yes':
    print("\n-------- E-BILL --------")
    for item, details in bill.items():
        print(f"{item.title()}  Qty:{details['quantity']}  Price:{details['price']}  Amount:{details['amount']}")
    print("------------------------")
    print("Total Amount:", total_amount)
    print("Thank you for shopping")
else:
    print("Thank you! Visit again")
```

```python
def show_items(price_list):
    print("Items available:")
    for item, price in price_list.items():
        print(item.title(), "-", price)


def take_order(price_list):
    order = {}
    while True:
        item = input("Enter item name: ").lower()

        if item not in price_list:
            print("Item not available")
            continue

        qty = int(input("Enter quantity: "))

        if item in order:
            order[item] += qty
        else:
            order[item] = qty

        more = input("Add more items? (yes/no): ").lower()
        if more != 'yes':
            break
    return order


def generate_bill(order, price_list):
    bill = {}
    total = 0
    for item, qty in order.items():
        amount = qty * price_list[item]
        bill[item] = {'qty': qty, 'price': price_list[item], 'amount': amount}
        total += amount
    return bill, total


def print_bill(bill, total):
    print("\n-------- E-BILL --------")
    for item, data in bill.items():
        print(item.title(), "Qty:", data['qty'],
              "Price:", data['price'],
              "Amount:", data['amount'])
    print("------------------------")
    print("Total Amount:", total)


def main():
    price_list = {'book': 50, 'pen': 10, 'pencil': 5}

    show_items(price_list)
    order = take_order(price_list)

    print("\nOrder taken successfully")
    print("Delivering your order")

    bill, total = generate_bill(order, price_list)

    choice = input("Do you want to print e-bill? (yes/no): ").lower()
    if choice == 'yes':
        print_bill(bill, total)
        print("Thank you for shopping")
    else:
        print("Thank you! Visit again")


main()
```

```python
class StationeryShop:

    def __init__(self):
        self.price_list = {
            'book': 50,
            'pen': 10,
            'pencil': 5
        }
        self.order = {}
        self.bill = {}
        self.total = 0

    def show_items(self):
        print("Items available:")
        for item, price in self.price_list.items():
            print(item.title(), "-", price)

    def take_order(self):
        while True:
            item = input("Enter item name: ").lower()

            if item not in self.price_list:
                print("Item not available")
                continue

            qty = int(input("Enter quantity: "))

            if item in self.order:
                self.order[item] += qty
            else:
                self.order[item] = qty

            more = input("Add more items? (yes/no): ").lower()
            if more != 'yes':
                break

        print("\nOrder taken successfully")
        print("Delivering your order")

    def generate_bill(self):
        for item, qty in self.order.items():
            amount = qty * self.price_list[item]
            self.bill[item] = {
                'qty': qty,
                'price': self.price_list[item],
                'amount': amount
            }
            self.total += amount

        print("\nBill generated successfully")

    def print_bill(self):
        print("\n-------- E-BILL --------")
        for item, data in self.bill.items():
            print(item.title(),
                  "Qty:", data['qty'],
                  "Price:", data['price'],
                  "Amount:", data['amount'])
        print("------------------------")
        print("Total Amount:", self.total)
        print("Thank you for shopping")


shop = StationeryShop()
shop.show_items()
shop.take_order()
shop.generate_bill()

choice = input("Do you want to print e-bill? (yes/no): ").lower()
if choice == 'yes':
    shop.print_bill()
else:
    print("Thank you! Visit again")
```

```python
def deco_outside(outside):
  def inner_func(a):
    a=10
    if type(a)==int or a%2==0:
      print("Even Integer: ")
      return outside(a)
    else:
      print("Odd Integer: ")
      return outside(a)
  return inner_func

```

```python
@deco_outside
def factorial(n):
    if n in(1,0) :
        return 1
    else:
        return n * factorial(n-1)
```
